**Converting Date Columns**

```python
df['Order Date']=pd.to_datetime(df['Order Date'])
df['Ship Date']=pd.to_datetime(df['Ship Date'])
```

```python
df.info()
```

```python
df.head()
```

**Adding New Date-Based Columns**

```python
df['Order Month']=df['Order Date'].dt.month
df['Order Year']=df['Order Date'].dt.year
df['Order Day of Week']=df['Order Date'].dt.dayofweek
```

```python
df.head()
```

**Descriptive Statistics**

```python
df.describe()
```

**Data Visualization**

**Monthly sales analysis**

```python
# Group sales by month
sales_by_month = df.groupby('Order Month')['Sales'].sum().reset_index()
sales_by_month
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(
    x='Order Month',
    y='Sales',
    data=sales_by_month,
    marker='o',
)
plt.title('Monthly Sales Analysis')
plt.show()
```

**Sales Analysis by Category**

```python
# Group sales by category
sales_by_category = df.groupby('Category')['Sales'].sum().reset_index()
sales_by_category
```

```python
# Extract values and labels
values = sales_by_category['Sales']
labels = sales_by_category['Category']

# Create pie chart
plt.figure(figsize=(10,8))
plt.pie(values,
        labels=labels,
        autopct='%1.1f%%'
)

plt.legend(labels,loc='best')
plt.title("Sales Analysis by Category")
plt.show()
```

**Sales Analysis by Sub-Category**

```python
sales_by_subcategory = df.groupby('Sub-Category')['Sales'].sum().reset_index()
sales_by_subcategory
```

```python
plt.figure(figsize=(18,8))
sns.barplot(x='Sub-Category', y='Sales',data=sales_by_subcategory)
plt.title("Sales Analysis by Sub-Category")
plt.show()
```

**Monthly Profit Analysis**

```python
profit_by_month = df.groupby('Order Month')['Profit'].sum().reset_index()
profit_by_month
```

```python
plt.figure(figsize=(10,5))
sns.lineplot(
    x='Order Month',
    y='Profit',
    data=profit_by_month,
    marker='o',
)
plt.title('Monthly Profit Analysis')
plt.show()
```

**Profit by Category**

```python
profit_by_category = df.groupby('Category')['Profit'].sum().reset_index()
profit_by_category
```

```python
# Extract values and labels
values = profit_by_category['Profit']
labels = profit_by_category['Category']

# Create pie chart
plt.figure(figsize=(10,8))
plt.pie(values,
        labels=labels,
        autopct='%1.1f%%'
)

plt.legend(labels,loc='best')
plt.title("Profit Analysis by Category")
plt.show()
```

**Profit by Sub-Category**

```python
profit_by_subcategory = df.groupby('Sub-Category')['Profit'].sum().reset_index()
profit_by_subcategory
```

```python
plt.figure(figsize=(18,8))
sns.barplot(x='Sub-Category', y='Profit',data=profit_by_subcategory)
plt.title("Profit Analysis by Sub-Category")
plt.show()
```

**Sales and Profit by Customer Segment**

```python
sales_profit_by_segment=df.groupby('Segment').agg({'Sales':'sum','Profit':'sum'}).reset_index()
sales_profit_by_segment
```

```python
df_melted = pd.melt(sales_profit_by_segment,
                    id_vars='Segment',
                    value_vars=['Sales', 'Profit'],
                    var_name='Metric',
                    value_name='Amount')

plt.figure(figsize=(10,5))
sns.barplot(x='Segment', y='Amount', hue='Metric', 
            hue_order=['Sales', 'Profit'],  # controls order
            data=df_melted)

plt.xlabel("Customer Segment")
plt.ylabel("Amount")
plt.title("Sales and Profit Analysis by Customer Segment")
plt.legend(title="Metric")
plt.show()
```

**Sales to Profit Ratio**

```python
sales_profit_by_segment = df.groupby('Segment').agg({'Sales': 'sum', 'Profit': 'sum'}).reset_index()
sales_profit_by_segment['Sales_to_Profit_Ratio'] = sales_profit_by_segment['Sales'] / sales_profit_by_segment['Profit']
print(sales_profit_by_segment[['Segment', 'Sales_to_Profit_Ratio']])
```
